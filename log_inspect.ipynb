```python
import json
from pathlib import Path

log_dir = Path("log")
latest_log = max(log_dir.glob("*.jsonl"), key=lambda p: p.stat().st_mtime, default=None)

if latest_log is None:
    raise FileNotFoundError("No JSONL files found in log/")

jsonl_file = Path(str(latest_log)).read_text()

logs = []
for line in jsonl_file.splitlines():
    try:
        log = json.loads(line)
        logs.append(log)
    except:
        pass
```

```python
import jmespath

selected_logs = logs

selected_logs = jmespath.search(
    "[?"
    'module==`"esp32_device_base"`'
    '&& command.parameters.value != `"hearbeat"`'
    ' && response.result.value != `"hearbeat"`'
    "]",
    selected_logs,
)

selected_logs
```
